```python
from web3 import Web3
import logging

# Enable debug logging
logging.basicConfig(level=logging.DEBUG)

# Your Geth HTTP RPC endpoint
rpc_url = "https://rpc1.fidesinnova.io:8545"

# Connect to Geth RPC
w3 = Web3(Web3.HTTPProvider(rpc_url))

# Check connection
if w3.is_connected():
    print("✅ Connected to Geth RPC")
    print(f"🔗 Current block number: {w3.eth.block_number}")
else:
    print("❌ Failed to connect to Geth RPC")
```
